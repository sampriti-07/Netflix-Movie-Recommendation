```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\DELL\Desktop\PROJECTS\Netflix-Movie-Recommendation\data\clean data\netflix_titles_clean.csv")
```

```python
df.head(5)
```

```python
df.info()
df.isnull().sum()
```

```python
df['type'].value_counts().plot(kind='bar')

plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```

moives makes large share in netflix

```python
df['rating'].value_counts()
```

```python
rating_counts = df['rating'].value_counts()

plt.plot(rating_counts.index, rating_counts.values, marker='o')

plt.xticks(ticks=range(len(rating_counts.index)),
           labels=rating_counts.index,
           rotation=90)

plt.title("Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.tight_layout()
plt.grid(True, linestyle='--')
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Country")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.show()
```

 The United States contributes the highest number of Netflix titles by a significant margin, far exceeding every other country in the dataset.


```python
df['director'].value_counts().head(10).plot(kind='line')

plt.title("Top 10 director")
plt.xticks(rotation=90)
plt.show()
```

A large number of titles have an unknown director, indicating that director information is missing for many entries

```python
df['release_year'].value_counts().sort_index().plot(figsize=(12,5))

plt.title("Netflix Content Released by Year")
plt.xlabel('released_year')
plt.ylabel('movie count')
plt.show()
```

contents release increased rapidly in 20's era

```python
movie_tv = df.groupby(['release_year', 'type']).size().unstack()

movie_tv.plot(figsize=(12,5))

plt.title("Movies and TV Shows Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Count")
plt.show()
```

movies consistently outnumbering TV shows and reaching a higher peak
