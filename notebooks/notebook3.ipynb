Select Required Columns

```python
features = ['director', 'cast', 'listed_in', 'description', 'country']

for col in features:
    df[col] = df[col].fillna('')
```

handle missing values

```python
df['tags'] = (
    df['listed_in'] + ' ' +
    df['director'] + ' ' +
    df['cast'] + ' ' +
    df['country'] + ' ' +
    df['description']
)
```

Create Tags Column

```python
df[['title', 'tags']].head()
```

```python
import re

def clean_text(text):
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    return text

df['tags'] = df['tags'].apply(clean_text)
```

```python
movies = df[['title', 'tags']]
```
